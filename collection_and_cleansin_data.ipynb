## Lấy dữ liệu từng mùa, từng vòng
`get_games_by_season(mùa, vòng)`

```python
seasons = []
for y in range(2017, 2025):
  seasons.append(f"{y}-{y+1}")
print(seasons)

match_rows, stat_rows = [],[]
missin_weeks = []

for season in seasons:
  for week in range(1,39):
    try:
      games = api.get_games_by_season(season, week)
    except ValueError as error:
      if "No games found" not in error:
        raise
      missin_weeks.append(season, week)
      continue
    for game in games:
      if game["league"]["name"] != "Premier League":
        continue
      match_rows.append({
          "game_id":game["id"],
          "season":game["season"],
          "match_week":game["match_week"],
          "date": game["date"],
          "home_team_id":game["home_team_id"],
          "away_team_id":game["away_team_id"],
          "home_team":game["home_team"]["name"],
          "away_team":game["away_team"]["name"],
          "home_goals":game['home_goals'],
          "away_goals":game['away_goals']
      }
      )
      stat_rows.extend(game.get('game_stats',[]))

colM = len(match_rows)
colS = len(stat_rows)
print("Số vòng thiếu:", len(missin_weeks))
print("Tổng số trận EPL:", colM)
print("Tổng số dòng thống kê của trận:", colS)
```

XEM THỬ BẢNG MATCHES

```python
matches=pd.DataFrame(match_rows)
display(matches.head())
```

XEM THỬ BẢNG STATS

```python
stats = pd.DataFrame(stat_rows)
display(stats.head())
```

## Làm sạch
1. Bỏ trùng theo khóa của trận và thống kê.
2. Chuẩn hóa tên đội và ngày thi đấu (str --> date)
3. Tỷ lệ cứu thua nằm ngoài **0–100%** là bất hợp lệ → chuyển thành `NaN`, **không** thay bằng 0 để tránh bias
4. Kiểm tra lại số dòng thống kê mỗi trận. (1 game_id chỉ được có 2 dòng trong bảng matches

```python
from re import S
matches = matches.drop_duplicates("game_id")
stats = stats.drop_duplicates(['game_id','team_id'])
print("Số dòng khác biệt khi bỏ duplicate:")
print("Bảng Matches",colM - len(matches))
print("Bảng Stats",colS - len(stats))
# đề phòng sau này thư viện có cập nhật thì code sẽ tự động loại bỏ

print()

matches['home_team'] = matches['home_team'].str.strip()
matches['away_team'] = matches['away_team'].str.strip()
# loại bỏ khoảng trang ở tên đội

matches['date'] = pd.to_datetime(matches['date'])
#chuyển từ string --> datetime

print("Kiểm tra giá trị invalid values")
tylequa100 = stats["save_percentage"]>100
print("Số dòng có tỷ lệ cứu thua vượt ngưỡng 100:",tylequa100.sum())
tyleam = stats["save_percentage"]<0
print("Số dòng có tỷ lệ dưới 0%:",tyleam.sum())

ty_le_sai = tylequa100 | tyleam
stats.loc[ty_le_sai,"save_percentage"] = float("nan")
# xử lí đưa về NAN để tránh bias
print()

trunglapdoi = matches["home_team_id"] == matches["away_team_id"]
if trunglapdoi.sum() > 0:
  raise ValueError(" Có trận đấu đội nhà trùng đội khách")

for game_id in matches["game_id"]:
  count = stats["game_id"] == game_id
  countTrue = stats[count]
  if len(countTrue) != 2:
    raise ValueError(f"Trận {game_id} không đủ 2 dòng stats")

print("-->Tổng số dòng có tỷ lệ không hợp lệ đã xử lý:",ty_le_sai.sum())
```
