```python
import sqlite3
import pandas as pd

matches = pd.read_csv('epl_matches_clean.csv')
stats = pd.read_csv('epl_team_stats_clean.csv')

conn = sqlite3.connect('epl.db')
matches.to_sql('matches', conn, if_exists='replace', index=False)
stats.to_sql('team_stats', conn, if_exists='replace', index=False)
conn.close()

print("Đã cất xong dữ liệu vào kho epl.db!")
```
